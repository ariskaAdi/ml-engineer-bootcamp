```python
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Python    :", sys.version.split()[0])
print("NumPy     :", np.__version__)
print("Pandas    :", pd.__version__)
print("Matplotlib:", matplotlib.__version__)
print("Lokasi    :", sys.executable)

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.title("Notebook success")
plt.show()

x = np.linspace(0, 10, 100)
data = np.random.default_rng(42).normal(170, 8, 500)   # 500 tinggi badan acak

fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))      # 1 baris, 3 grafik

axes[0].plot(x, np.sin(x))
axes[0].set_title("Line — tren")

axes[1].bar(["Kopi", "Teh", "Roti"], [28.8, 11.7, 12.3])
axes[1].set_title("Bar — perbandingan")

axes[2].hist(data, bins=20)
axes[2].set_title("Histogram — sebaran")

plt.tight_layout()
plt.show()
```
